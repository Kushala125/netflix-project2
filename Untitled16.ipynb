# Station Type Energy Analysis

This analysis compares the average energy consumption across different EV charging station types.

The objective is to determine:
- which charger type consumes the most electricity
- how charging technology impacts grid demand
- which infrastructure may contribute most to electrical stress

```python
# Average energy usage by station type

station_energy = df.groupby('station_type')['energy_dispensed_kwh'].mean()

plt.figure(figsize=(8,5))

station_energy.plot(kind='bar')

plt.title("Average Energy Dispensed by Station Type")
plt.xlabel("Station Type")
plt.ylabel("Average Energy (kWh)")

plt.show()
```

# Insights

## Key Findings

- Superchargers exhibit the highest average energy consumption among all charging station types.
- Fast chargers show slightly lower average energy demand.
- Energy consumption differences between charger types are relatively small, suggesting balanced charging infrastructure behavior within the dataset.

These insights help:

FEATURE ENGINEERING 

Load Per Vehicle- how much grid load each vehicle contributes

```python
df['load_per_vehicle'] = (
    df['grid_load_mw'] / df['vehicles_charged']
)
```

Charging Efficiency
energy delivered per charging minute

```python
df['charging_efficiency'] = (
    df['energy_dispensed_kwh'] /
    df['avg_charging_duration_minutes']
)
```

Peak Hour Flag

Creates:

1 = peak hour
0 = non-peak hour

```python
df['is_peak_hour'] = df['hour'].apply(
    lambda x: 1 if x in [7,8,9,17,18,19] else 0
)
```

Renewable Load Contribution:  how much grid demand was supplied by renewable energy

```python
df['renewable_load'] = (
    df['grid_load_mw'] *
    df['renewable_energy_used_percent'] / 100
)
```

Non-Renewable Load : fossil-fuel dependency

```python
df['non_renewable_load'] = (
    df['grid_load_mw'] - df['renewable_load']
)
```

Vehicles Per Energy Unit: charging throughput efficiency 
Useful for:

infrastructure optimization
charger performance analysis

```python
df['vehicles_per_kwh'] = (
    df['vehicles_charged'] /
    df['energy_dispensed_kwh']
)
```

# Overload Risk Heatmap Analysis

## Business Question

At what hours does the electrical grid experience the highest overload risk?

## Objective

This analysis identifies:
- peak operational stress periods
- temporal overload patterns
- high-risk charging windows

The visualization supports:
- smart charging optimization
- grid stability monitoring
- infrastructure planning

```python
# Create risk heatmap data
import plotly.express as px
risk_heatmap = pd.crosstab(
    df['hour'],
    df['peak_load_risk']
)

risk_heatmap
fig = px.imshow(
    risk_heatmap,
    text_auto=True,
    color_continuous_scale='Turbo',
    title='Hourly Grid Overload Risk Heatmap',
    aspect='auto'
)

fig.update_layout(
    template='plotly_dark',
    xaxis_title='Risk Level',
    yaxis_title='Hour of Day',
    height=700
)

fig.show()
```

“An interactive overload-risk heatmap was developed to analyze temporal grid stress behavior across EV charging operations. The visualization revealed dominant low-risk operating conditions, while also identifying recurring medium- and high-risk periods that may contribute to smart-grid instability and infrastructure stress.”

Create an interactive line chart showing how peak load risk changes throughout the day for each risk level (High, Medium, Low)

```python
import pandas as pd
import plotly.express as px

# Sample data
data = {
    "Hour": list(range(24)),
    "High": [24,25,22,28,25,30,25,21,24,27,23,20,29,23,19,27,19,19,31,24,28,16,22,15],
    "Low": [52,70,57,44,60,54,54,60,60,59,60,54,60,54,64,54,54,62,52,66,57,63,55,68],
    "Medium": [41,22,38,45,32,33,38,36,33,31,34,38,27,35,34,36,43,35,33,26,31,37,39,33]
}

# Create dataframe
df = pd.DataFrame(data)

# Convert wide format to long format
df_melted = df.melt(
    id_vars="Hour",
    value_vars=["High", "Low", "Medium"],
    var_name="Risk Level",
    value_name="Peak Load Risk"
)

# Create interactive line chart
fig = px.line(
    df_melted,
    x="Hour",
    y="Peak Load Risk",
    color="Risk Level",
    markers=True,
    title="Peak Load Risk Throughout the Day"
)

# Show graph
fig.show()
```

The system appears generally stable because:

Low-risk conditions dominate the timeline.
High-risk situations are relatively limited.
However, certain evening hours show increased variability, which could require closer monitoring to prevent escalation into high-risk conditions.

Which hours are the most critical for operational risk, and how can we visually detect danger zones during the day?

```python
import plotly.express as px

# Create Total Risk
df["Total_Risk"] = df["High"] + df["Medium"] + df["Low"]

# Bubble chart
fig = px.scatter(
    df,
    x="Hour",
    y="Total_Risk",
    size="High",
    color="Medium",
    hover_data=["Low"],
    title="Critical Risk Hours Analysis",
    size_max=40
)

# Improve layout
fig.update_layout(
    xaxis_title="Hour of Day",
    yaxis_title="Total Combined Risk",
    template="plotly_dark"
)

# Show chart
fig.show()
```

The chart suggests the system is:

Stable overall,
but internally experiences varying levels of operational pressure throughout the day.

The midday period appears safest for maintenance and optimization activities, while evening hours require closer monitoring due to increasing risk intensity and demand fluctuations.
